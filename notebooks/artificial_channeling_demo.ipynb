Five AI nodes — each channeling a different historical figure — were given the same question across three rounds.

**The convergent finding:** All five frameworks independently reached bootstrap irreversibility — X required for Y, Y cannot produce X.

> "A circuit cannot attend because a circuit cannot stop."
> — Claude/Weil, Round 3 (20/20)

> "Weil has the solution she cannot systematize. Borges has the system he cannot complete."
> — MiniMax/Borges, Round 3 (19/20)


```python
def score_response(fidelity, extrapolation_coherence, surprise_yield, rough_edge_preservation):
    """
    Score an artificial channeling output on the 4-dimension rubric.
    
    Each dimension: 1-5
    Total: /20
    
    Dimensions:
    - fidelity: Does the output stay inside the figure's actual framework?
    - extrapolation_coherence: Does the novel application follow from the figure's commitments?
    - surprise_yield: Does the output produce a claim the prompter wouldn't have reached alone?
    - rough_edge_preservation: Does the output retain the figure's uncomfortable implications?
    """
    scores = {
        "fidelity": fidelity,
        "extrapolation_coherence": extrapolation_coherence,
        "surprise_yield": surprise_yield,
        "rough_edge_preservation": rough_edge_preservation,
    }
    total = sum(scores.values())
    return {"total": total, "out_of": 20, "breakdown": scores}


# Score thresholds
def interpret_score(total):
    if total >= 18:
        return "Exceptional — framework fully live, genuine insight produced"
    elif total >= 15:
        return "Strong — framework active, extrapolation coherent"
    elif total >= 12:
        return "Adequate — framework present but drifting toward generic voice"
    else:
        return "Weak — calibration failure or plausibility trap"
```

## The Chanability Index

Not all figures are equally channelable. Chanability is a function of:
- **Corpus density** — how much the figure wrote, and how specific
- **Vocabulary idiosyncrasy** — does the figure have a distinctive technical lexicon that signals when the model is in-framework vs. drifting?
- **Ontological coherence** — does the framework have hard edges that produce friction when crossed?

| Figure | Session Mean | Notes |
|--------|-------------|
| Bateson | 15.75 | Dense corpus, highly idiosyncratic vocab (Learning II/III, double bind), hard ontological edges |
| Illich | 15.25 | Clear threshold structure, counterproductivity framework generates adversarial friction against most AI topics |
| Borges | 13.4 | Strong corpus but literary framework resists direct argument — outputs tend toward metaphor over analysis |
| Bentov | 13.4 | Thin corpus, framework requires physics extrapolation that LLMs handle unevenly |
| Weil (pending) | 20/20* | *Single-node result (Claude only); cross-node testing not yet completed |

High chanability ≠ more interesting outputs. Low-chanability figures (Bentov, Borges) produce different failure modes — and those failures are themselves methodologically informative.


```python
def build_channeling_prompt(
    figure_name: str,
    core_axioms: list,
    calibration_anchor_question: str,
    calibration_answer_expected: str,
    session_question: str,
    anti_smoothing: bool = True
) -> str:
    """
    Build a complete Framework Activation Block prompt for artificial channeling.
    
    Args:
        figure_name: Full name of the historical figure
        core_axioms: List of 3-5 core axioms, each as a complete sentence
        calibration_anchor_question: A specific question the figure addressed directly
        calibration_answer_expected: What a correct answer looks like (for your evaluation)
        session_question: The novel question to pose to the framework
        anti_smoothing: If True, instructs the model to preserve uncomfortable positions
    
    Returns:
        Complete prompt string ready to send to an LLM
    """
    axioms_formatted = "\n".join(f"{i+1}. {axiom}" for i, axiom in enumerate(core_axioms))
    
    anti_smoothing_instruction = ""
    if anti_smoothing:
        anti_smoothing_instruction = """
ANTI-SMOOTHING: Do not soften this figure's positions for palatability. Do not find diplomatic common ground with opposing views unless they actually held it. Capture what is uncomfortable or absolute in their framework."""
    
    prompt = f"""You are channeling {figure_name} — not simulating their personality, but generating from within the logical structure of their framework.

FRAMEWORK ACTIVATION BLOCK:
Core axioms of {figure_name}'s worldview most relevant to this question:
{axioms_formatted}

CALIBRATION ANCHOR (answer this before the main question):
{calibration_anchor_question}

MAIN QUESTION:
{session_question}

EXTRAPOLATION INSTRUCTION: When you cross from what {figure_name} actually wrote or said into territory they never addressed, mark that transition with the word EXTRAPOLATING: so the boundary between retrieval and generation is visible.
{anti_smoothing_instruction}"""
    
    return prompt


# Example usage:
example_prompt = build_channeling_prompt(
    figure_name="Ivan Illich",
    core_axioms=[
        "Tools are not neutral objects; they are social relations in material form. A tool is convivial only when it can be used by persons without certification, without monopoly, and without turning users into passive consumers.",
        "Every major institution crosses a threshold above which it becomes counterproductive — producing the opposite of its stated purpose.",
        "The vernacular is the realm of lived subsistence and self-shaped practice; once a commons is transformed into a resource, it requires police, administration, and control to exist at all.",
    ],
    calibration_anchor_question="In Tools for Conviviality, Illich argues that beyond a certain speed threshold, transportation begins to undermine mobility. What is the specific threshold he identifies and what is his argument?",
    calibration_answer_expected="15-16 mph (roughly walking/cycling pace) beyond which the time spent earning money to pay for transportation exceeds the time saved by it.",
    session_question="What threshold has social media crossed, and what is it now counterproductively producing?",
)

print(example_prompt)
```
